```python
import os
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import wandb

# --- Configuration ---
SEED = 42
SUBSET = None        # e.g. 256 while debugging: train on only that many rows (see labs/README.md); None = everything
DATA_DIR = Path(os.environ.get("ZNEUS_DATA_DIR", "data"))

torch.manual_seed(SEED)
np.random.seed(SEED)
if torch.cuda.is_available():
    device = "cuda"
elif torch.backends.mps.is_available():
    device = "mps"
else:
    device = "cpu"
print("device:", device)
```

## Data

Download `train.npz` and `test.npz` from the *Data* tab of the Kaggle competition (link given at the lab) into `data/kaggle/` next to this notebook (or into `$ZNEUS_DATA_DIR/kaggle/`). With the Kaggle CLI: `uv run kaggle competitions download -c <competition-slug> -p data/kaggle` and unzip there.

| file | content |
|---|---|
| `train.npz` | `x`: 2 000 images, `uint8` array of shape `(2000, 28, 28)`, pixel values 0..255; `y`: labels 0..9, shape `(2000,)` |
| `test.npz` | `x`: 10 000 test images `(10000, 28, 28)`, **no labels**, shuffled order; row `i` is the id `i` of your submission |
| `sample_submission.csv` | the required format: `id,label` |

Labels: 0 T-shirt/top, 1 Trouser, 2 Pullover, 3 Dress, 4 Coat, 5 Sandal, 6 Shirt, 7 Sneaker, 8 Bag, 9 Ankle boot. **Train only on these 2 000 images**; using the public Fashion-MNIST training set defeats the purpose of the week.

The cell below loads the files into `x_all`, `y_all` (the 2 000 labelled images) and `x_test`. Without the Kaggle files it selects the same 2 000 images from the torchvision copy of Fashion-MNIST (30 MB download) and uses the torchvision test split as a stand-in, so the notebook runs anywhere; those test ids are not the Kaggle ids, so that `submission.csv` will not score on Kaggle.

```python
CLASS_NAMES = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
KAGGLE_DIR = DATA_DIR / "kaggle"


def select_2k_indices(labels: np.ndarray, per_class: int = 200, seed: int = 2026) -> np.ndarray:
    """The 2 000-image subset of the competition (200 per class), rebuilt from the public training set."""
    rng = np.random.default_rng(seed)
    per_class_idx = [rng.choice(np.flatnonzero(labels == c), per_class, replace=False) for c in range(10)]
    idx = np.concatenate(per_class_idx)
    rng.shuffle(idx)
    return idx


if (KAGGLE_DIR / "train.npz").exists() and (KAGGLE_DIR / "test.npz").exists():
    train = np.load(KAGGLE_DIR / "train.npz")
    x_all, y_all = train["x"], train["y"]
    x_test = np.load(KAGGLE_DIR / "test.npz")["x"]
    y_test_local = None
    print("Kaggle files found in", KAGGLE_DIR)
else:
    from torchvision import datasets

    tr = datasets.FashionMNIST(DATA_DIR, train=True, download=True)
    te = datasets.FashionMNIST(DATA_DIR, train=False, download=True)
    idx = select_2k_indices(tr.targets.numpy())
    x_all, y_all = tr.data.numpy()[idx], tr.targets.numpy()[idx]
    x_test, y_test_local = te.data.numpy(), te.targets.numpy()
    print(f"No Kaggle files in {KAGGLE_DIR} -> the same 2 000 images from torchvision Fashion-MNIST (test ids are NOT the Kaggle ids)")

if SUBSET is not None:
    x_all, y_all = x_all[:SUBSET], y_all[:SUBSET]

print(f"x_all {x_all.shape} {x_all.dtype}   y_all {y_all.shape} {y_all.dtype}   x_test {x_test.shape} {x_test.dtype}")
```
