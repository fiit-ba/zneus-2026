# Week 1 · PyTorch basics (optional self-study)

Weeks 2 and 3 build a small neural-network framework from raw tensors and assume you are comfortable with tensors, broadcasting, reductions, `@` versus `*`, NumPy interop, devices and autograd. Rather than reinventing a tensor tutorial, we point you to the classics. Pick one from the first group and work through it before week 2.

## PyTorch first (learn the framework itself)

- **[Official PyTorch tutorials: Learn the Basics](https://docs.pytorch.org/tutorials/beginner/basics/intro.html)** and the **[Deep Learning with PyTorch: A 60 Minute Blitz](https://docs.pytorch.org/tutorials/beginner/deep_learning_60min_blitz.html)**. The shortest path: tensors, autograd, `nn.Module`, and a CIFAR10 classifier. Read this before the week 2 lab.
- **[Learn PyTorch for Deep Learning: Zero to Mastery](https://www.learnpytorch.io/)** by Daniel Bourke. A free online book with 10 notebooks, plus a [26-hour video version on freeCodeCamp's YouTube](https://www.freecodecamp.org/news/learn-pytorch-for-deep-learning-in-day/). The most beginner-friendly option. [GitHub repository](https://github.com/mrdbourke/pytorch-deep-learning).

## University-level, PyTorch-based

- **[UvA Deep Learning Tutorials](https://uvadlc-notebooks.readthedocs.io/)** by Phillip Lippe (University of Amsterdam, MSc AI). Notebooks from PyTorch basics through optimization, transformers, GNNs, normalizing flows and Vision Transformers, with recorded sessions. [GitHub repository](https://github.com/phlippe/uvadlc_notebooks).
- **[Deep Learning Fundamentals](https://lightning.ai/courses/deep-learning-fundamentals/)** by Sebastian Raschka. 10 units of short videos with quizzes and exercises, all in PyTorch, from logistic regression to self-attention and LLMs. [Code and exercises](https://github.com/Lightning-AI/dl-fundamentals).

## Broader classics

- **[Dive into Deep Learning](https://d2l.ai/)**. Free textbook with a PyTorch code track, used by many universities.
- **[fast.ai Practical Deep Learning for Coders](https://course.fast.ai/)**. Top-down and project-driven, built on PyTorch.
- **[Neural Networks: Zero to Hero](https://karpathy.ai/zero-to-hero.html)** by Andrej Karpathy. Builds autograd and GPT from scratch in plain Python and PyTorch. Excellent for intuition, not a framework tutorial.
- **[Deep Learning with PyTorch](https://www.manning.com/books/deep-learning-with-pytorch)** by Stevens, Antiga and Viehmann. Manning made the PDF free.
